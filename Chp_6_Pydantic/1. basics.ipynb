Pydantic

```
from pydantic import BaseModel


class User(BaseModel):
    name:str
    email:str
    age:int
    password:str
```

```
user1 = User(**{
    "name": "Mehek Jain",
    "email": "mehek@gmail.com",
    "age": 22,
    "password": "mypassword123"})

user1
```

```
user1 = User( name = "Mehek Jain", email = "mehek@gmail.com", age =  22, password = "mypassword123")
user1
```

## **Coercion VS Strict Types**

### Coercion

```
user1 = User( name = "Mehek Jain", email = "mehek@gmail.com", age =  "22", password = "mypassword123") #passed string for int. It tries to convert the data type to defined data type
user1
```

### Strict Types

```
from pydantic import StrictInt


class User(BaseModel):
    name:str
    email:str
    age:StrictInt
    password:str

```

```
user1 = User( name = "Mehek Jain", email = "mehek@gmail.com", age =  "22", password = "mypassword123") #throws error in strict types
user1  
```
