```
from pydantic import BaseModel, Field, EmailStr, field_validator


class User(BaseModel):
    name:str = Field(..., min_length=3, max_length=50)
    email:EmailStr = Field(..., description="User email address")
    age:int = Field(...,  gt = 18, strict=True)
    password:str = Field(..., min_length=8, max_length=20)


    @field_validator("password")
    @classmethod
    def password_validator(cls, value):
        special_char = ["@", "#", "$","%","&","*"]
        if not any(char.isdigit() for char in value):
            raise ValueError("Password must have a digit")

        if not any(char in special_char for char in value):
            raise ValueError("Atleast 1 special character is required")

        return value
```

```
user1 = User( name = "Mehek Jain", email = "mehek@gmail.com", age =  22, password = "mypassword@123") #throws error in strict types
user1
```
